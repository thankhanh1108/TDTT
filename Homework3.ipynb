```python
h=int(input())
if h<=0:
    print("Du lieu ko hop le")
else:
    for i in range(1,h+1):
        letrai=(" "*(h-i))
        if i==h:
            print((2*i-1)*"*")
        elif i==1:
            print(letrai+"*")
        else:
            giua=" "*(2*i-3)
            print(letrai+"*"+giua+"*")
```

```python
n=int(input())
if n<=0:
    print("Error")
else:
    songuyento=0
    while n>0:
        uoc=0
        for i in range(1,(n%10)+1):
            if (n%10)%i==0:
                uoc+=1
        if uoc==2:
            songuyento+=1
        n//=10
    print(songuyento)
```
